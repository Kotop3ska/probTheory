```python
import random
from collections import Counter
import matplotlib.pyplot as plt
```

```python
# Параметры симуляции
N = 10_0000  # количество экспериментов
awakenings = []  # список всех пробуждений

for _ in range(N):
    coin = random.choice(['Orel', 'Reshka'])
    if coin == 'Orel':
        awakenings.append('Orel')
    else:
        awakenings.append('Reshka')  # 1 день
        awakenings.append('Reshka')  # 2 день

counts = Counter(awakenings)
total = len(awakenings)
prob_reshka = counts['Reshka'] / total

print(f"Всего пробуждений: {total}")
print(f"Орел: {counts['Orel']} ({counts['Orel']/total:.1%})")
print(f"Решка: {counts['Reshka']} ({prob_reshka:.4f} или {prob_reshka:.1%})")

# График
plt.figure(figsize=(8, 6))
plt.bar(counts.keys(), counts.values(), color=['skyblue', 'salmon'], alpha=0.8, edgecolor='black')
plt.title('Симуляция пробуждений Спящей красавицы (N=100000)', fontsize=14)
plt.xlabel('Исход монеты')
plt.ylabel('Количество пробуждений')
plt.ylim(0, max(counts.values()) * 1.1)

# Аннотации
for i, (outcome, count) in enumerate(counts.items()):
    plt.text(i, count + total*0.002, f'{count}\n({count/total:.1%})', ha='center', va='bottom')

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
